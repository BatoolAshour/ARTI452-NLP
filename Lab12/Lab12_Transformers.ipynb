## 1. Load a Transformer

We use **DistilBERT**, a  Transformer fine-tuned for positive/negative sentiment classification.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForSequenceClassification.from_pretrained(model_name)
model.eval()

# Look inside the Transformer architecture.
print("Transformer layers:", model.config.n_layers)
print("Attention heads per layer:", model.config.n_heads)
print("Hidden size:", model.config.dim)
```


<img src="https://drive.google.com/thumbnail?id=1v2dRK1KR_ydZOknjj8Xv8_wjOdXleDYs&sz=w1000" width="500">

## 2. Tokenizing the text

```python
sentence = "The movie was not good."

# Tokenization: converts text into token IDs
inputs = tokenizer(sentence, return_tensors="pt")

# [CLS] = start token, [SEP] = end token
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

print("Tokens:", tokens)
```

## 3. Embeddings and self-attention

Each token ID becomes a **vector (embedding)**. Inside each Transformer layer, **self-attention** lets a token use information from other tokens in the same sentence.

```python
import torch

# Enable attention weights
model.set_attn_implementation("eager")

with torch.no_grad():
    embeddings = model.distilbert.embeddings(inputs["input_ids"])
    outputs = model(**inputs, output_attentions=True)

# Attention from the first layer and first head
attention = outputs.attentions[0][0, 0]

print("Embeddings shape:", embeddings.shape)
print("Attention shape:", attention.shape)
print("Tokens:", tokens)
print("Attention weights:\n", attention)
```

## 4. Classification: From Transformer to prediction

The Transformer produces contextual representations. The **classification head** turns them into two scores (*logits*). **Softmax** converts these into probabilities that sum to 1.

```python
# Get prediction scores from the Transformer
with torch.no_grad():
    logits = model(**inputs).logits

# Convert scores into probabilities
probs = torch.softmax(logits, dim=1)[0]

# Display the prediction
prediction = probs.argmax().item()

print("Prediction:", model.config.id2label[prediction])
print("Confidence:", round(probs[prediction].item(), 3))
```

### 1. Load `bert-base-uncased`

```python
from transformers import AutoTokenizer, AutoModel
import torch

bert_name = "bert-base-uncased"
bert_tokenizer = AutoTokenizer.from_pretrained(bert_name)
bert_model = AutoModel.from_pretrained(bert_name, attn_implementation="eager")  # eager -> attention weights available
bert_model.eval()
```

### 2. Number of layers and attention heads per layer

```python
num_layers = bert_model.config.num_hidden_layers
num_heads = bert_model.config.num_attention_heads

print("Transformer layers:", num_layers)          # 12
print("Attention heads per layer:", num_heads)    # 12
```

### 3. Total attention heads across all layers

```python
total_heads = num_layers * num_heads
print("Total attention heads:", total_heads)      # 12 x 12 = 144
```

### 4. Attention weights: 3rd layer, 2nd head

```python
sentence = "Transformers use self-attention to understand language."

bert_inputs = bert_tokenizer(sentence, return_tensors="pt")
bert_tokens = bert_tokenizer.convert_ids_to_tokens(bert_inputs["input_ids"][0])

with torch.no_grad():
    bert_outputs = bert_model(**bert_inputs, output_attentions=True)

# bert_outputs.attentions: tuple of 12 tensors, each [batch, heads, seq_len, seq_len]
attention_l3_h2 = bert_outputs.attentions[2][0, 1]

print("Tokens:", bert_tokens)
print("Attention shape:", attention_l3_h2.shape)   # [seq_len, seq_len]
print("Attention weights (layer 3, head 2):\n", attention_l3_h2)
print("Each row sums to 1:", attention_l3_h2.sum(dim=-1))
```

 Heatmap to see it more clearly:

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 6))
plt.imshow(attention_l3_h2, cmap="viridis")
plt.xticks(range(len(bert_tokens)), bert_tokens, rotation=90)
plt.yticks(range(len(bert_tokens)), bert_tokens)
plt.xlabel("Key tokens (attended TO)")
plt.ylabel("Query tokens (attending FROM)")
plt.title("BERT attention — Layer 3, Head 2")
plt.colorbar()
plt.tight_layout()
plt.show()
```

### 5. What the rows and columns mean

The matrix is **`seq_len × seq_len`** — one row and one column per token (including `[CLS]` and `[SEP]`).

- **Rows = query tokens**: the token that is *paying attention* (attending **from**).
- **Columns = key tokens**: the tokens that *receive attention* (attended **to**).
- **Cell `[i, j]`** = how much token *i* attends to token *j* when building its new representation in this layer/head.
- Each **row sums to 1** (softmax), so a row is a probability distribution showing how token *i* splits its attention across the whole sentence.

**Example:** a high value in the row for `"attention"` under the column for `"self"` means that, in layer 3 / head 2, the model mixes in a lot of information from `"self"` when representing `"attention"`
